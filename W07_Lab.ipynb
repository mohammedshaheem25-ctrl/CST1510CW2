## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
raw_line = "  Srv-01 , used=87 , total=100  "

# 1. .strip() the whole line
line = raw_line.strip()

# 2. Split off the hostname into hostname_part and rest
hostname_part, rest = line.split(",", 1)

# 3. Clean the hostname
hostname = hostname_part.strip().lower()

# 4 & 5. Build the dictionary by parsing the rest
data = {"hostname": hostname}

for piece in rest.split(","):
    key, value = piece.split("=")
    data[key.strip()] = float(value.strip())

print(data)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
def parse_line(raw_line):
    """Parse a raw server status line into a dictionary."""
    parts = raw_line.split(",")
    hostname = parts[0].lower()
    
    result = {"hostname": hostname}
    for part in parts[1:]:
        key, val = part.split("=")
        result[key] = float(val)
        
    return result

print(parse_line("SRV-02,used=120,total=100"))
````

````python
"""
RECORD CHECK  -  my version
===========================

Name  : mohammed shaheem
Lane  : cyber security 
Date  : 11/11/2026

Run it:   python template.py

Work through the numbered sections in order. Each one tells you what it must do.
Delete these instructions as you replace them with your code.
"""

# =================================================================== FUNCTIONS
# 1. Write parse_line(raw_line) that cleans and parses one raw, messy line
#    into a dict. Handle inconsistent spacing and case - that is the whole
#    point of this week.
#
#    Typical and above: also write status_of(...), shaped for your lane
#    (e.g. OVER LIMIT / WARNING / OK, or FLAGGED / OK). Give both a
#    one-line docstring.

def parse_line(raw_line):
    """Parse a raw server status line with inconsistent spacing and case into a dictionary."""
    parts = [p.strip() for p in raw_line.split(",")]
    hostname = parts[0].lower()
    
    result = {"hostname": hostname}
    for part in parts[1:]:
        key, val = part.split("=")
        result[key.strip().lower()] = float(val.strip())
        
    return result

def status_of(data):
    """Determine the status of a server based on its used versus total capacity ratio."""
    ratio = data["used"] / data["total"]
    if ratio >= 1.0:
        return "OVER LIMIT"
    elif ratio >= 0.9:
        return "WARNING"
    else:
        return "OK"


# ====================================================================== INPUT
# 2. Your raw, messy records - at least 5, deliberately varied in spacing
#    and case. Match the shape to your lane:
#
#    IT                  : "   Srv-01 , used=87 , total=100   "
#
#    Excellent: save these lines to a file and load them with open() /
#    a for loop over the file, instead of writing them directly here.

# Let's write messy sample records to a file first to meet the 'Excellent' criteria
messy_data = """   SRV-01 , used=87 , total=100   
srv-02,USED=120,TOTAL=100
   SRV-03  ,  used=45  ,  total=100
srv-04, used=95 , total = 100 
   SRV-05 , used=200 , total=150   
"""

with open("messy_records.txt", "w") as f:
    f.write(messy_data)


# =============================================================== PROCESS + OUTPUT
# 3. Parse and report on every record.
#
#    Threshold : parse each record, print the cleaned dict.
#    Typical   : also call status_of(...) for each, print it alongside.
#    Excellent : print a one-line summary at the end - how many records,
#                how many flagged.

total_records = 0
flagged_records = 0

with open("messy_records.txt", "r") as f:
    for raw_record in f:
        raw_record = raw_record.strip()
        if not raw_record:
            continue
            
        total_records += 1
        
        try:
            cleaned_dict = parse_line(raw_record)
            server_status = status_of(cleaned_dict)
            
            if server_status in ("OVER LIMIT", "WARNING"):
                flagged_records += 1
                
            print(f"{cleaned_dict} -> {server_status}")
        except ValueError:
            print(f"Could not parse: {raw_record}")

print(f"\nSummary: {total_records} records processed, {flagged_records} flagged.")


# ==========================================================================
# 4. Before you finish:
#
#    [X] Run it against at least 5 genuinely messy raw records
#    [X] Run it with a value that will not convert to a number - write the
#        error down (ValueError: could not convert string to float)
#    [X] Check every function does one job, with a docstring
````
